```python
import pandas as pd
import numpy as np

# Read the CSV file
df = pd.read_csv("Sushil_Prose_1777833444.csv")

# Basic info
print(f"Total Keystrokes: {len(df)}")
print(f"Columns: {df.columns.tolist()}")
print(f"User Name: {df['User_Name'].unique()[0]}")
print(f"Session Type: {df['Session_Type'].unique()[0]}\n")

# Dwell Time Stats
dwell_mean = df['Dwell_Time'].mean() * 1000 # convert to ms
dwell_std = df['Dwell_Time'].std() * 1000
print(f"Dwell Time - Mean: {dwell_mean:.2f} ms | Std Dev: {dwell_std:.2f} ms")

# Flight Time Stats (Excluding Outliers)
valid_flights = df[df['Outlier'] == False]['Flight_Time']
flight_mean = valid_flights.mean() * 1000
flight_std = valid_flights.std() * 1000
print(f"Flight Time - Mean: {flight_mean:.2f} ms | Std Dev: {flight_std:.2f} ms")

# Outliers check
outliers_count = df['Outlier'].sum()
print(f"Total Outliers (Pauses > 2s): {outliers_count}")

# Check unique keys (to see if modifiers were successfully ignored)
unique_keys = df['Key'].unique()
print(f"\nUnique Keys Sample (First 20): {unique_keys[:20]}")

# Look for 'shift', 'ctrl', 'alt'
bad_keys = [k for k in unique_keys if type(k) == str and k.lower() in ['shift', 'ctrl', 'alt', 'caps lock']]
print(f"Forbidden Modifiers Found: {bad_keys}")
```
